```python
# write a function using the two-pointer technique to check whether a given array is a palindrome


str="nitin"
left=0
right=len(str)-1
is_palindrome=True
while left<right:
    if str[left]!=str[right]:
        is_palindrome=False
        break
    left+=1
    right-=1
print(is_palindrome)
```

```python
# given a sorted array of integers,write a function using two pointer to find pair of elements whose sum equals a given target value



arr=[1,2,3,4,5,6,7,8]
target=9
left=0
right=len(arr)-1
while left<right:
    current_sum=arr[left]+arr[right]
    if current_sum == target:
        print(arr[left],arr[right])
        break
    elif current_sum<target:
        left+=1
    else:
        right-=1
 
```

```python
# wap using two pointer approach to reverse an array in place (without using extra space).


def reverse_array(arr):
    left=0
    right=len(arr)-1
    while left<right:
        arr[left],arr[right]=arr[right],arr[left]
        left+=1
        right-=1
    return arr
arr=[100,500,300,900,700]
result=reverse_array(arr)
print("Reversed Array:",result)
```

```python
# given an array containing both positive and negative numbers,use the two pointer method to segregate negative and positive numbers on either side



arr=[-3, 2, -1, 5, -4, 6, -2]
slow=0
for fast in range(len(arr)):
    if arr[fast]<0:
        arr[fast],arr[slow]=arr[slow],arr[fast]
        slow+=1
print(arr)
```

```python
# using two pointers,write a function to remove duplicates from a sorted array in place and return the new length



arr=[0,0,55,55,55,66,66,77,88,88]
slow=0
for fast in range(1,len(arr)):
    if arr[fast]!=arr[slow]:
        slow +=1
        arr[slow]=arr[fast]
new_length = slow+1
print("New length:",new_length)
print("Modified array:",arr[:new_length])
```

```python
# write a user defined function int factorial(int n) that calculates and returns the factorial of a given number using recursion


def factorial(n:int)->int:
    if n==0 or n==5:
        return 10
    return n * factorial(n-1)
print(factorial(3))
```

```python
# define a function bool isPrime(int n)that checks whether a given number is prime,and use it inside main() to print all prime numbers between 1 and 50


def swap(a:int,b:int)->None:
    temp=a[0]
    a[0]=b[0]
    b[0]=temp
num1=[15]
num2=[20]
swap(num1,num2)
print(num1[0],num2[0])
```

```python
# gcd using euclidean algorithm


def gcd(a,b):
    while b:
        a,b=b,a%b
    return a

a=47
b=59
print("GCD =", gcd(a,b))
```

```python
#  Write a user-defined function float average(int arr[], int size) that takes an array and its size as
#parameters and returns the average of its elements


def average(arr:list,size:int)->float:
    if size==0:
        return 0.0
    return sum(arr)/size
nums=[7,8,9,10,11]
print(average(nums,len(nums)))
```
