```python
#Question9
def prime(n):
    if n <= 1:
        return False
    if n <= 3:
        return True
    if n %2 ==0 or n %3 ==0 or n % 5 ==0:
        return False
    if n % n ==0:
        return True

for n in range (1,30):
    if prime(n):
        print(f"{n} is a prime number")
    else:
        print(f"{n} is not a prime number")
```

```python
#Question10
def reverse_and_slice(text):
    reversed_text = text[::-1]
    first_half = text[:len(text)//2]
    return reversed_text, first_half


result = reverse_and_slice("Python")
print(result)
```
